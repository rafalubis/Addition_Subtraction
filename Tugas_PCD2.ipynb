```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
```

```python
img_upnvj = cv2.imread('upnvj.jpg', cv2.IMREAD_GRAYSCALE)
img_cr7 = cv2.imread('cr7.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
h_upnvj, w_upnvj = img_upnvj.shape
h_cr7, w_cr7 = img_cr7.shape
```

```python
min_h = min(h_upnvj, h_cr7)
min_w = min(w_upnvj, w_cr7)
```

```python
gray_upnvj = img_upnvj[0:min_h, 0:min_w]
gray_cr7 = img_cr7[0:min_h, 0:min_w]
```

```python
img_addition = cv2.add(gray_upnvj, gray_cr7)
img_subtraction = cv2.subtract(gray_upnvj, gray_cr7)
```

```python

print("1. Citra grayscale upnvj :")
cv2_imshow(gray_upnvj)

print("2. Citra grayscale cr7:")
cv2_imshow(gray_cr7)

print("3. Citra addition :")
cv2_imshow(img_addition)

print("4. Citra subtractuin :")
cv2_imshow(img_subtraction)
```
